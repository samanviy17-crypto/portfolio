## Popcorn Hack 1

For a 324-square-foot practice area, I predict an 18.0-foot side. The required jersey range is 1–55. The supplied game scores average 24.4, so rounding upward gives 25. Picking a score or jersey again can repeat a previous result.

```python
# CODE_RUNNER: Popcorn Hack 1

import math
import random
print("randint documentation:", random.randint.__doc__)
print("ceil documentation:", math.ceil.__doc__)
points_per_game = [24, 18, 31, 27, 22]
selected_game = random.choice(points_per_game)
jersey = random.randint(1, 55)
side = math.sqrt(324)
average = sum(points_per_game) / len(points_per_game)
print("Game to review:", selected_game)
print("Jersey:", jersey)
print("Court side:", side)
print("Average:", average, "Rounded up:", math.ceil(average))
assert selected_game in points_per_game and 1 <= jersey <= 55
assert side == 18.0 and math.ceil(average) == 25
```

## Popcorn Hack 2

The roster uses fictional players Nora, Elias, and June. My added `/record` route reports wins and losses. Routes are defined before creating the test client; no server process is needed.

```python
# CODE_RUNNER: Popcorn Hack 2

from flask import Flask, jsonify
app = Flask(__name__)

@app.route("/score")
def score():
    return jsonify({"home": 62, "away": 59, "quarter": 4})

@app.route("/roster")
def roster():
    return jsonify({"players": ["Nora", "Elias", "June"]})

@app.route("/record")
def record():
    return jsonify({"wins": 4, "losses": 2})

with app.test_client() as client:
    for route in ["/score", "/roster", "/record"]:
        response = client.get(route)
        assert response.status_code == 200
        print(route, "->", response.get_json())
    assert client.get("/record").get_json() == {"wins": 4, "losses": 2}
    assert client.get("/missing").status_code == 404
```

## Homework Hack — field hockey season report

These fictional games record goals in the required `points` field. One scoreless match tests whether zero remains in the average: 7 goals over four games gives 1.75. Cedar appears twice and accounts for 6 goals. The best match has 4. My next fixture is November 7, 2026; its countdown changes with the execution date. An empty season needs an explicit guard because `statistics.mean([])` raises an exception.

```python
# CODE_RUNNER: Homework Hack — field hockey season report

import json
import statistics
from datetime import date

def summarize_season(season):
    """Summarize field hockey scoring.

    Parameters:
        season: list of game dictionaries containing opponent (str) and
            points (nonnegative number); optional date metadata is preserved
            in the caller's data, which this function does not modify.
    Returns:
        dict with total_points, average_points, best_game (highest point
        value or None if empty), and points_by_opponent (accumulated totals).
        An empty season has total 0 and average 0.0.
    """
    points = []
    points_by_opponent = {}
    for game in season:
        points.append(game["points"])
        opponent = game["opponent"]
        points_by_opponent[opponent] = points_by_opponent.get(opponent, 0) + game["points"]
    return {
        "total_points": sum(points),
        "average_points": round(statistics.mean(points), 2) if points else 0.0,
        "best_game": max(points) if points else None,
        "points_by_opponent": points_by_opponent
    }

games = [
    {"date": "2026-09-20", "opponent": "Cedar", "points": 2},
    {"date": "2026-09-23", "opponent": "Orchard", "points": 0},
    {"date": "2026-09-26", "opponent": "Cedar", "points": 4},
    {"date": "2026-09-29", "opponent": "Pine", "points": 1}
]
report = summarize_season(games)
next_game = date(2026, 11, 7)
report["report_date"] = date.today().isoformat()
report["next_game"] = next_game.isoformat()
report["days_until_next_game"] = (next_game - date.today()).days
print(json.dumps(report, indent=2))
assert report["total_points"] == 7 and report["average_points"] == 1.75
assert report["best_game"] == 4 and report["points_by_opponent"]["Cedar"] == 6
print("requirements.txt: pandas and Flask; standard library modules need no installation")
```

## Tests and dependency file

The assignment’s [requirements.txt]({{site.baseurl}}/assets/python-homework/libraries/requirements.txt) lists Flask and pandas for the imagined full app. `json`, `datetime`, and `statistics` ship with Python and are excluded.
```python
# CODE_RUNNER: Tests and dependency file

import json
import statistics
from datetime import date

def summarize_season(season):
    """Summarize field hockey scoring.

    Parameters:
        season: list of game dictionaries containing opponent (str) and
            points (nonnegative number); optional date metadata is preserved
            in the caller's data, which this function does not modify.
    Returns:
        dict with total_points, average_points, best_game (highest point
        value or None if empty), and points_by_opponent (accumulated totals).
        An empty season has total 0 and average 0.0.
    """
    points = []
    points_by_opponent = {}
    for game in season:
        points.append(game["points"])
        opponent = game["opponent"]
        points_by_opponent[opponent] = points_by_opponent.get(opponent, 0) + game["points"]
    return {
        "total_points": sum(points),
        "average_points": round(statistics.mean(points), 2) if points else 0.0,
        "best_game": max(points) if points else None,
        "points_by_opponent": points_by_opponent
    }

assert summarize_season([]) == {"total_points": 0, "average_points": 0.0, "best_game": None, "points_by_opponent": {}}
games = [{"opponent": "Orchard", "points": 5}, {"opponent": "Orchard", "points": 7}]
before = json.dumps(games)
result = summarize_season(games)
assert result == {"total_points": 12, "average_points": 6, "best_game": 7, "points_by_opponent": {"Orchard": 12}}
assert json.dumps(games) == before
assert (date(2026, 11, 7) - date(2026, 9, 30)).days == 38
print("Empty season:", summarize_season([]))
print("Repeated opponent:", result)
print("Fixed-date countdown test: 38 days")
```
