## Popcorn Hack 1

Prediction: each original draw is one of 1, 2, 3, 4; I cannot predict the exact random sequence. Both endpoints are possible and repeated values are allowed. Changing the upper bound to 3 removes only record 4.

```python
# CODE_RUNNER: Popcorn Hack 1

import random
for upper_bound in [4, 3]:
    observed = []
    for trial in range(8):
        record_number = random.randint(1, upper_bound)
        assert 1 <= record_number <= upper_bound
        observed.append(record_number)
    print("Inclusive bounds: 1 to", upper_bound, "Observed:", observed)
```

## Popcorn Hack 3

There are nine possible record/action pairs. GET should show a search, POST a creation-validation request, and PUT an update. After random runs, explicit action choices verify all branches instead of assuming randomness covers them.

```python
# CODE_RUNNER: Popcorn Hack 3

import random
parts = [
    {"product_name": "Replacement Flywheels", "category": "Auto Racing", "spec_number": "1.1"},
    {"product_name": "Multiple Disc Clutch Assemblies", "category": "Drag Racing", "spec_number": "1.2"},
    {"product_name": "Racing Flywheel Record", "category": "Auto Racing", "spec_number": "2.1"}
]

def preview_action(part, action):
    spec = part["spec_number"]
    if action == "GET/search":
        return f"Search requested for spec {spec}"
    elif action == "POST/create":
        return f"Validate spec {spec} before creation"
    elif action == "PUT/update":
        return f"Update requested for spec {spec}"
    raise ValueError("Unknown action")

actions = ["GET/search", "POST/create", "PUT/update"]
for trial in range(4):
    selected_part = random.choice(parts)
    selected_action = random.choice(actions)
    print(selected_part["product_name"], selected_action, "->", preview_action(selected_part, selected_action))
for action in actions:
    print("Forced action:", preview_action(parts[1], action))
```

## Homework Hack — backend QA simulator

Each run chooses one intact record and one of four actions. Creation adds a new spec, repeated creation rejects it, deletion removes it, and missing searches/updates/deletes return not found. Prediction: every event uses an allowed pair and reports the result against the current simulated state.

```python
# CODE_RUNNER: Homework Hack — backend QA simulator

import random
parts = [
    {"product_name": "Replacement Flywheels", "category": "Auto Racing", "spec_number": "1.1"},
    {"product_name": "Multiple Disc Clutch Assemblies", "category": "Drag Racing", "spec_number": "1.2"},
    {"product_name": "Racing Flywheel Record", "category": "Auto Racing", "spec_number": "2.1"}
]

actions = ["GET/search", "POST/create", "PUT/update", "DELETE/remove"]

def handle_action(part, action, stored_specs):
    """Simulate one action against a local set; never contact a real backend."""
    spec = part["spec_number"]
    if action == "GET/search":
        return "found" if spec in stored_specs else "not found"
    elif action == "POST/create":
        if spec in stored_specs:
            return "rejected duplicate"
        stored_specs.add(spec)
        return "created"
    elif action == "PUT/update":
        return "updated" if spec in stored_specs else "not found"
    elif action == "DELETE/remove":
        if spec in stored_specs:
            stored_specs.remove(spec)
            return "removed"
        return "not found"
    raise ValueError("Unknown action")

def simulate_qa(test_count, existing_spec_numbers, rng=None):
    """Run test_count trials with fresh local state and return the event log."""
    if rng is None:
        rng = random.Random()
    stored_specs = set(existing_spec_numbers)
    events = []
    for trial in range(test_count):
        part = rng.choice(parts)
        action = rng.choice(actions)
        result = handle_action(part, action, stored_specs)
        event = (part["spec_number"], action, result)
        events.append(event)
        print(trial + 1, part["product_name"], "|", *event)
    print("Final stored specs:", sorted(stored_specs))
    return events

existing_spec_numbers = ["1.1", "2.1"]
test_count = 5
events = simulate_qa(test_count, existing_spec_numbers)
assert len(events) == test_count
assert existing_spec_numbers == ["1.1", "2.1"]
```

## Tests and pseudocode trace

Three records × four actions gives 12 possible selected pairs; repeats remain valid. For AP recordIndex 2 and actionNumber 2, Python selects `parts[1]` and POST/create: spec 1.2 is initially absent, so it is created; repeating that action rejects the duplicate. Seeded debug runs repeat exactly, while the normal homework runner is unseeded.

```python
# CODE_RUNNER: Tests and pseudocode trace

import random
parts = [
    {"product_name": "Replacement Flywheels", "category": "Auto Racing", "spec_number": "1.1"},
    {"product_name": "Multiple Disc Clutch Assemblies", "category": "Drag Racing", "spec_number": "1.2"},
    {"product_name": "Racing Flywheel Record", "category": "Auto Racing", "spec_number": "2.1"}
]

actions = ["GET/search", "POST/create", "PUT/update", "DELETE/remove"]

def handle_action(part, action, stored_specs):
    """Simulate one action against a local set; never contact a real backend."""
    spec = part["spec_number"]
    if action == "GET/search":
        return "found" if spec in stored_specs else "not found"
    elif action == "POST/create":
        if spec in stored_specs:
            return "rejected duplicate"
        stored_specs.add(spec)
        return "created"
    elif action == "PUT/update":
        return "updated" if spec in stored_specs else "not found"
    elif action == "DELETE/remove":
        if spec in stored_specs:
            stored_specs.remove(spec)
            return "removed"
        return "not found"
    raise ValueError("Unknown action")

def simulate_qa(test_count, existing_spec_numbers, rng=None):
    """Run test_count trials with fresh local state and return the event log."""
    if rng is None:
        rng = random.Random()
    stored_specs = set(existing_spec_numbers)
    events = []
    for trial in range(test_count):
        part = rng.choice(parts)
        action = rng.choice(actions)
        result = handle_action(part, action, stored_specs)
        event = (part["spec_number"], action, result)
        events.append(event)
        print(trial + 1, part["product_name"], "|", *event)
    print("Final stored specs:", sorted(stored_specs))
    return events

# Enumerate every pair against fresh initial state, independent of random coverage.
expected = {
    "1.1": ["found", "rejected duplicate", "updated", "removed"],
    "1.2": ["not found", "created", "not found", "not found"],
    "2.1": ["found", "rejected duplicate", "updated", "removed"]
}
for part in parts:
    for index, action in enumerate(actions):
        actual = handle_action(part, action, {"1.1", "2.1"})
        assert actual == expected[part["spec_number"]][index]
        print("Pair test:", part["spec_number"], action, actual)
# Verify dependent actions, including creating the same spec twice.
stored = {"1.1", "2.1"}
for action, expected_result in [("POST/create", "created"), ("POST/create", "rejected duplicate"),
                                ("DELETE/remove", "removed"), ("GET/search", "not found")]:
    result = handle_action(parts[1], action, stored)
    assert result == expected_result
    print("State test:", action, result)
first = simulate_qa(5, ["1.1", "2.1"], random.Random(23))
second = simulate_qa(5, ["1.1", "2.1"], random.Random(23))
assert first == second
assert simulate_qa(0, ["1.1", "2.1"], random.Random(23)) == []
print("All 12 pairs and state-change tests passed")
```
