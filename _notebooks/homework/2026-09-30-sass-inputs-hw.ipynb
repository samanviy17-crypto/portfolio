## Practice — Read a Value and Submit

This separate demonstration extends the required markup with a local submit interaction. Try an empty value, spaces, and a skill such as event setup. It does not submit information to PVO.

```python
%%html
<!-- UI_RUNNER: Inputs Value and Submit Practice | editor: true, id: inputs-value-and-submit-practice -->
<div id="preview-inputs-value-and-submit-practice">
<form>
 <label for="inputs-skill">Volunteer skill</label>
 <input id="inputs-skill" name="skill" type="text" class="ocs__input medium" placeholder="For example, event setup">
 <button type="submit" class="ocs__btn accent fill">Preview volunteer skill</button>
 <p aria-live="polite" data-value>Current value: (empty)</p>
 <output aria-live="polite">Enter a skill to preview it here.</output>
</form>
</div>
<script>
(function () {
const root = document.getElementById("preview-inputs-value-and-submit-practice");
const input = root.querySelector('input');
const live = root.querySelector('[data-value]');
const result = root.querySelector('output');
input.addEventListener('input', () => { live.textContent = 'Current value: ' + (input.value || '(empty)'); });
root.querySelector('form').addEventListener('submit', event => {
 event.preventDefault();
 const skill = input.value.trim();
 result.textContent = skill ? 'Volunteer skill: ' + skill : 'Please enter a volunteer skill.';
});
})();
</script>
```

## Explanation and pseudocode

HTML supplies labels, fields, and a submit button. SASS supplies reusable appearance. JavaScript reads `.value` and responds to events. AP CSP pseudocode describes the same input/output algorithm without browser details.

<pre><code>skill ← INPUT()
IF (skill = "")
{
    DISPLAY("Please enter a volunteer skill.")
}
ELSE
{
    DISPLAY(skill)
}</code></pre>

Input values are strings: with `rawAge = "17"`, `rawAge + 1` produces `"171"`, while `Number(rawAge) + 1` produces `18`. Empty strings need validation before conversion because `Number("")` is zero; nonnumeric text produces NaN. For a simple email check, `volunteer@example.org` passes a basic @/domain check, while an empty value or `volunteer` fails. A basic pattern cannot prove that a mailbox exists.

The displayed value uses textContent, so typed HTML is displayed as text. Labels identify the fields even after placeholders disappear. The large, medium, small, and gradient modifiers replace hardcoded dimensions and colors.
