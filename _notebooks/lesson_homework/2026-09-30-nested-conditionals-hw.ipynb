## Popcorn Hack 1

For weekday True and homework False, the outer weekday branch and inner unfinished branch print Finish homework. A finished weekday prints Free time; a weekend skips the homework check.

```python
# CODE_RUNNER: Popcorn Hack 1

for is_weekday, homework_finished in [(True, False), (True, True), (False, False)]:
    print("Weekday:", is_weekday, "Finished:", homework_finished)
    if is_weekday:
        if homework_finished:
            print("Free time")
        else:
            print("Finish homework")
    else:
        print("Weekend plan")
```

Weekday: False Finished: False
Weekend plan</pre>

## Popcorn Hack 2

Two separate top-level `if` statements check permission even when the student is logged out. Indenting the permission check inside the login branch makes it dependent on successful login.

```python
# CODE_RUNNER: Popcorn Hack 2

for logged_in, has_permission in [(True, True), (True, False), (False, True), (False, False)]:
    print("Login:", logged_in, "Permission:", has_permission)
    if logged_in:
        print("Account found")
        if has_permission:
            print("Page opened")
        else:
            print("Permission needed")
    else:
        print("Log in first")
```

## Popcorn Hack 3

The outer lunch decision runs first; only then does its practice check run. The four Boolean combinations produce four distinct recommendations.

```python
# CODE_RUNNER: Popcorn Hack 3

def lunch_plan(lunch_packed, practice_after_school):
    if lunch_packed:
        if practice_after_school:
            return "Eat packed lunch and bring a practice snack"
        else:
            return "Eat packed lunch; no practice snack needed"
    else:
        if practice_after_school:
            return "Buy lunch and an extra practice snack"
        else:
            return "Buy lunch; no practice snack needed"

answers = set()
for lunch_packed in [True, False]:
    for practice_after_school in [True, False]:
        answer = lunch_plan(lunch_packed, practice_after_school)
        print(lunch_packed, practice_after_school, "->", answer)
        answers.add(answer)
assert len(answers) == 4
```

True False -&gt; Eat packed lunch; no practice snack needed
False True -&gt; Buy lunch and an extra practice snack
False False -&gt; Buy lunch; no practice snack needed</pre>

## Homework Hack

The account decision runs before any schedule information is used. Homework completion selects a branch before sports scheduling is considered, and lunch is checked only inside the logged-in path. This keeps a logged-out student from receiving a personalized schedule.

```python
# CODE_RUNNER: Homework Hack

def daily_dashboard(student):
    if student["logged_in"]:
        if student["homework_finished"]:
            if student["practice_today"]:
                activity = "Homework done: go to practice, then dinner"
            else:
                activity = "Homework done: enjoy free time before dinner"
        else:
            if student["practice_today"]:
                activity = "Finish homework before practice, then dinner"
            else:
                activity = "Finish homework before dinner"
        if student["lunch_packed"]:
            meal = "Eat your packed lunch"
        else:
            meal = "Buy lunch at school"
        return activity + ". " + meal + "."
    else:
        return "Log in first to see your daily plan."

student = {"logged_in": True, "homework_finished": False, "lunch_packed": True, "practice_today": True}
print(daily_dashboard(student))
```

### Recorded output

<pre>Finish homework before practice, then dinner. Eat your packed lunch.</pre>

## Tests

All 16 Boolean combinations cover every meaningful path. Logged-out cases share one answer; the eight logged-in combinations yield distinct activity/meal recommendations.

```python
# CODE_RUNNER: Tests

def daily_dashboard(student):
    if student["logged_in"]:
        if student["homework_finished"]:
            if student["practice_today"]:
                activity = "Homework done: go to practice, then dinner"
            else:
                activity = "Homework done: enjoy free time before dinner"
        else:
            if student["practice_today"]:
                activity = "Finish homework before practice, then dinner"
            else:
                activity = "Finish homework before dinner"
        if student["lunch_packed"]:
            meal = "Eat your packed lunch"
        else:
            meal = "Buy lunch at school"
        return activity + ". " + meal + "."
    else:
        return "Log in first to see your daily plan."

answers = set()
for logged_in in [False, True]:
    for homework_finished in [False, True]:
        for lunch_packed in [False, True]:
            for practice_today in [False, True]:
                student = dict(logged_in=logged_in, homework_finished=homework_finished,
                               lunch_packed=lunch_packed, practice_today=practice_today)
                result = daily_dashboard(student)
                print(student, "->", result)
                assert bool(result)
                if not logged_in:
                    assert result == "Log in first to see your daily plan."
                answers.add(result)
assert len(answers) == 9
```
