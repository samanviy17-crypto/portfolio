## Popcorn

Warm-up: searching for 15 uses 5 linear checks and 3 binary checks, saving 2. Searching for 3 uses 1 and 3, saving -2. Checkpoint prediction: n=8 gives 8 linear and 4 binary worst-case checks; n=16 gives 16 and 5. Binary search has better worst-case growth for a sorted catalog, but does not win every individual search.

```python
# CODE_RUNNER: Popcorn

import math
for linear_checks, binary_checks in [(5, 3), (1, 3)]:
    print("Linear:", linear_checks, "Binary:", binary_checks, "Saved:", linear_checks - binary_checks)
for n in [8, 16]:
    linear_checks = n
    binary_checks = math.floor(math.log2(n)) + 1 if n > 0 else 0
    print("Books:", n)
    print("Worst-case linear:", linear_checks)
    print("Worst-case binary:", binary_checks)
```

## Homework

Prediction: 12 books → 12 visits and 144 ordered pairs; 24 → 24 visits and 576 pairs. Doubling n doubles n but multiplies n² by four. These are different tasks; scanning cannot replace pairing when pairs are required.

```python
# CODE_RUNNER: Homework

for n in [12, 24]:
    visits = n
    ordered_pairs = n * n
    print("Books:", n, "Visits:", visits, "Ordered pairs:", ordered_pairs)
    assert ordered_pairs == n ** 2
```

## Tests

The main, doubled, and empty models all start fresh. For repeated searches in a sorted 1,000-entry catalog, choose binary search: 10 worst-case inspections instead of 1,000. These formulas count modeled work; they do not time or execute a search.

```python
# CODE_RUNNER: Tests

import math
for n, expected in [(12, (12, 12, 144)), (24, (24, 24, 576)), (0, (0, 0, 0))]:
    actual = n, n, n * n
    print("Books, visits, pairs:", actual)
    assert actual == expected
for n, expected in [(0, 0), (1, 1), (8, 4), (16, 5), (1000, 10)]:
    binary_checks = math.floor(math.log2(n)) + 1 if n > 0 else 0
    print("Catalog:", n, "Linear worst case:", n, "Binary worst case:", binary_checks)
    assert binary_checks == expected
```

## Design Thinking

The librarian needs dependable lookups as the catalog grows. My goal is to compare counts with readable labels. I considered a linear scan and binary search and built a calculator for both. Testing an empty catalog confirmed the zero guard avoids log2(0).

For unsorted data, linear search works immediately; binary search first needs sorting, which costs time. Keeping a separate sorted copy also uses memory proportional to the number of entries. Repeated queries can justify that preparation, whereas a single query might not. A short-books-first reading heuristic can fit many books into a budget but may miss the most enjoyable combination.
