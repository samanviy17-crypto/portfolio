## Popcorn Hack — Fill in the Blanks

1. **Checkbox** stores whether the toggle is on or off.
2. **Slider** is the visible part that changes color.
3. **Alignment** is one of the switch wrapper’s layout responsibilities.
4. **border-radius** rounds the slider’s edges.
5. **White circle** is the thumb created by the before pseudo-element.

More precisely, `::before` is a pseudo-element, not a class. The public OCS track class provides the visible switch; the hidden checkbox holds its state.

## Popcorn Practice — Find the Bug

The missing class is `ocs__toggle-track`; aria-hidden keeps the decorative track out of the accessibility tree.

```python
%%html
<!-- UI_RUNNER: Toggles Find the Bug | editor: true, id: toggles-find-the-bug -->
<div id="preview-toggles-find-the-bug">
<label class="ocs__toggle ocs__toggle--green">
 <input id="toggles-bug" class="ocs__toggle-input" type="checkbox">
 <span class="ocs__toggle-track" aria-hidden="true"></span>
 <span class="ocs__toggle-label">Enable notifications</span>
</label>
<p data-confirmation aria-live="polite"></p>
<output aria-live="polite">Notifications: off</output>
</div>
<script>
(function () {
const root = document.getElementById("preview-toggles-find-the-bug");
root.querySelector('[data-confirmation]').textContent = root.querySelector('.ocs__toggle-track') ? 'Toggle works! 🎉' : 'Toggle track is missing.';
const toggle = root.querySelector('input');
const status = root.querySelector('output');
toggle.addEventListener('change', () => { status.textContent = 'Notifications: ' + (toggle.checked ? 'on' : 'off'); });
})();
</script>
```

## Challenge — Connect a Feature

```python
%%html
<!-- UI_RUNNER: Toggles Connected Message | editor: true, id: toggles-connected-message -->
<div id="preview-toggles-connected-message">
<label class="ocs__toggle ocs__toggle--green">
 <input id="toggles-message" class="ocs__toggle-input" type="checkbox">
 <span class="ocs__toggle-track" aria-hidden="true"></span>
 <span class="ocs__toggle-label">Show study tip</span>
</label>
<p id="toggles-tip" hidden>Practice one small example, then explain why it works.</p>
</div>
<script>
(function () {
const root = document.getElementById("preview-toggles-connected-message");
const toggle = root.querySelector('input');
const tip = root.querySelector('#toggles-tip');
// Initialize from checked as well as listening for future changes.
const update = () => { tip.hidden = !toggle.checked; };
toggle.addEventListener('change', update);
update();
})();
</script>
```

## Homework — Study Controls

Three independent switches show notes, enable a practice input, and show a break reminder. Their labels describe actual changes. The live count is derived from current checkbox states.

```python
%%html
<!-- UI_RUNNER: Toggles Homework | editor: true, id: toggles-homework -->
<div id="preview-toggles-homework">
<label class="ocs__toggle ocs__toggle--green">
 <input id="toggles-notes" class="ocs__toggle-input" type="checkbox">
 <span class="ocs__toggle-track" aria-hidden="true"></span>
 <span class="ocs__toggle-label">Show study notes</span>
</label>
<label class="ocs__toggle ocs__toggle--green">
 <input id="toggles-practice" class="ocs__toggle-input" type="checkbox">
 <span class="ocs__toggle-track" aria-hidden="true"></span>
 <span class="ocs__toggle-label">Enable practice answer</span>
</label>
<label class="ocs__toggle ocs__toggle--green">
 <input id="toggles-reminder" class="ocs__toggle-input" type="checkbox">
 <span class="ocs__toggle-track" aria-hidden="true"></span>
 <span class="ocs__toggle-label">Show break reminder</span>
</label>
<section id="toggles-notes-panel" hidden><h3>Study notes</h3><p>A checkbox stores a Boolean state; a change listener applies it to a feature.</p></section>
<section><h3>Practice question</h3><label for="toggles-answer">Which property reports whether a toggle is on?</label><input id="toggles-answer" class="ocs__input medium" type="text" placeholder="Enable practice answer first" disabled></section>
<p id="toggles-reminder-panel" hidden>Break reminder: pause and stretch after a focused study session.</p>
<output id="toggles-count" aria-live="polite">Toggles on: 0 / 3</output>
</div>
<script>
(function () {
const root = document.getElementById("preview-toggles-homework");
// Scope queries to this runner so other exercises do not affect the count.
const notes = root.querySelector('#toggles-notes');
const practice = root.querySelector('#toggles-practice');
const reminder = root.querySelector('#toggles-reminder');
const toggles = [notes, practice, reminder];
function update() {
 // checked is a Boolean, while value would be a string unrelated to the on/off state.
 root.querySelector('#toggles-notes-panel').hidden = !notes.checked;
 // Disabling the input changes a real feature, not just the switch's appearance.
 root.querySelector('#toggles-answer').disabled = !practice.checked;
 root.querySelector('#toggles-reminder-panel').hidden = !reminder.checked;
 // Recalculate from current state to avoid an incorrect count after repeated clicks.
 const on = toggles.filter(toggle => toggle.checked).length;
 root.querySelector('#toggles-count').textContent = 'Toggles on: ' + on + ' / 3';
}
toggles.forEach(toggle => toggle.addEventListener('change', update));
// Initialize output on every Run, before the user changes anything.
update();
})();
</script>
```

## Explanation and test cases

The unchecked initial state hides both optional messages and disables the answer field. Each change recomputes all features and the count. Test counts 0 → 1 → 2 → 3, then turn switches off in a different order and return to zero. Use Tab and Space as well as pointer clicks. A toggle suits a persistent on/off preference; a one-time submit action should use a button. Four explanatory comments plus an initialization comment document the state, scope, counting, and feature behavior.
