## Exercise 1 — descriptive names

I renamed the broad make/mix/bake steps to prepare_vanilla_cupcakes, measure_dry_ingredients, combine_batter, and bake_cupcakes. They describe the purpose without opening every function.

```python
# CODE_RUNNER: Exercise 1 — descriptive names

def measure_dry_ingredients():
    print("Measure 2 cups flour and 1 cup sugar")

def combine_batter():
    print("Mix dry ingredients with eggs and milk")

def bake_cupcakes():
    print("Bake cupcakes at 350 F for 20 minutes")

def prepare_vanilla_cupcakes():
    measure_dry_ingredients()
    combine_batter()
    bake_cupcakes()

prepare_vanilla_cupcakes()
```

## Naming mini-game

Editable variables replace keyboard prompts so Run works on the website. Try prepare_dessert and do_step; both execute the same sequence, but only the first communicates the goal. Blank names use a default; cancellation skips execution.

```python
# CODE_RUNNER: Naming mini-game

import random

def play_game(procedure_name, command):
    name = procedure_name.strip() or "prepare_ingredient"
    def prepare_dessert():
        ingredient = random.choice(["flour", "sugar", "eggs", "milk"])
        print(f"Mixing {ingredient} into the bowl")
        print("Heating the oven")
        print("Dessert is ready!")
    print("Procedure name:", name)
    if command.strip().lower() == "run":
        prepare_dessert()
        return True
    print("Canceled")
    return False

assert play_game("prepare_dessert", "run")
assert play_game("do_step", "run")
assert not play_game("", "cancel")
```

Canceled</pre>

## Exercise 3 — decomposition and tests

I split recipe planning into scaling, temperature conversion, and coordination. Each pure helper can be tested without running the whole recipe. Zero batches and an invalid negative count test boundaries.

```python
# CODE_RUNNER: Exercise 3 — decomposition and tests

def scale_recipe(batches):
    """Return flour cups and sugar cups for a nonnegative number of batches."""
    if batches < 0:
        raise ValueError("Batches cannot be negative")
    return {"flour_cups": 2 * batches, "sugar_cups": batches}

def fahrenheit_to_celsius(fahrenheit):
    return (fahrenheit - 32) * 5 / 9

def plan_cupcakes(batches):
    return {"ingredients": scale_recipe(batches), "oven_celsius": round(fahrenheit_to_celsius(350), 1)}

assert scale_recipe(0) == {"flour_cups": 0, "sugar_cups": 0}
assert scale_recipe(2) == {"flour_cups": 4, "sugar_cups": 2}
assert fahrenheit_to_celsius(32) == 0
assert fahrenheit_to_celsius(212) == 100
try:
    scale_recipe(-1)
except ValueError as error:
    print("Negative test:", error)
else:
    raise AssertionError("Negative batches must fail")
print("Two-batch plan:", plan_cupcakes(2))
print("Helper and integration tests passed")
```
