## Popcorn Hack 2

The original score 9 reaches the first `>= 3` branch and misleadingly says it is probably fine. The `>= 6` and `>= 9` branches are unreachable because every matching value was already consumed. Testing higher thresholds first repairs the chain.

```python
# CODE_RUNNER: Popcorn Hack 2

for risk_score in [0, 3, 6, 9]:
    print("Score:", risk_score)
    # 3 thresholds -> 4 outcomes -> 4 answers
    if risk_score >= 9:
        print("Do not reply; verify the request through a trusted channel.")
    elif risk_score >= 6:
        print("High risk; ask a trusted person to help check.")
    elif risk_score >= 3:
        print("Some warning signs; verify before responding.")
    else:
        print("No strong warning signs from these checks.")
```

## Homework Hack

Prediction: the supplied message triggers all signals and scores 9. Independent scoring `if` statements accumulate evidence; their missing `else` branches intentionally add nothing. The final reporting chain always supplies an answer.

```python
# CODE_RUNNER: Homework Hack

def advise(message, amount_requested, known_contact):
    risk_score = 0
    # No else needed: absence of this signal adds zero to the accumulator.
    if "gift card" in message.lower():
        risk_score += 4
    # No else needed: a smaller amount leaves the accumulated score alone.
    if amount_requested > 100:
        risk_score += 3
    # No else needed: a known sender adds no unknown-contact penalty.
    if not known_contact:
        risk_score += 2
    print("Risk score:", risk_score)
    # 3 thresholds -> 4 outcomes -> 4 answers
    if risk_score >= 9:
        print("Do not reply; verify the request through a trusted channel.")
    elif risk_score >= 6:
        print("High risk; ask a trusted person to help check.")
    elif risk_score >= 3:
        print("Some warning signs; verify before responding.")
    else:
        print("No strong warning signs from these checks.")
    return risk_score

message = "URGENT: send gift cards now to unlock your account"
amount_requested = 500
known_contact = False
assert advise(message, amount_requested, known_contact) == 9
```

### Recorded output

<pre>Risk score: 9
Do not reply; verify the request through a trusted channel.</pre>

## Tests

Both extremes and the two intermediate outcomes run from fresh scores. Exactly 100 does not earn the over-100 increment.

```python
# CODE_RUNNER: Tests

def advise(message, amount_requested, known_contact):
    risk_score = 0
    # No else needed: absence of this signal adds zero to the accumulator.
    if "gift card" in message.lower():
        risk_score += 4
    # No else needed: a smaller amount leaves the accumulated score alone.
    if amount_requested > 100:
        risk_score += 3
    # No else needed: a known sender adds no unknown-contact penalty.
    if not known_contact:
        risk_score += 2
    print("Risk score:", risk_score)
    # 3 thresholds -> 4 outcomes -> 4 answers
    if risk_score >= 9:
        print("Do not reply; verify the request through a trusted channel.")
    elif risk_score >= 6:
        print("High risk; ask a trusted person to help check.")
    elif risk_score >= 3:
        print("Some warning signs; verify before responding.")
    else:
        print("No strong warning signs from these checks.")
    return risk_score

for message, amount, known, expected in [
    ("See you at lunch", 0, True, 0),
    ("Please send money", 101, True, 3),
    ("Send gift cards", 0, False, 6),
    ("Send gift cards", 500, False, 9),
    ("Please send money", 100, True, 0)
]:
    assert advise(message, amount, known) == expected
```
