## Popcorn Hack — Why Cats Purr

A heading identifies the topic, a subordinate heading introduces the explanation, and a paragraph contains the supporting text. `ocs_lead` marks its introductory role.

```python
%%html
<!-- UI_RUNNER: Typography Popcorn Base | editor: true, id: typography-popcorn-base -->
<div class="ocs_card">
 <h2 class="ocs_section-title">Why Cats Purr</h2>
 <h3>A Sign of Happiness</h3>
 <p class="ocs_lead">Cats usually purr when they feel safe, cozy, and ready for a nap.</p>
</div>
```

## Homework Hack — How to Pat a Cat

The page title already supplies the top-level heading, so this component begins with h2. Its sections use h3. A real ordered list supplies numbering and conveys sequence to assistive technology.

```python
%%html
<!-- UI_RUNNER: Typography Homework Base | editor: true, id: typography-homework-base -->
<div class="ocs_card">
 <h2 class="ocs_section-title">How to Pat a Cat</h2>
 <h3>Have you ever wanted to...</h3>
 <p class="ocs_lead">pat a cat? Well good for you: here are step-by-step instructions on how to pat a cat!</p>
 <p class="ocs_text"><strong>How to pat a cat:</strong></p>
 <ol>
  <li>Put out your hand and let the cat approach.</li>
  <li>Touch the cat gently if it welcomes contact.</li>
  <li>Move your hand back and forth gently; stop if the cat moves away.</li>
 </ol>
 <h3>Congrats! You pat a cat :D</h3>
 <p class="ocs_text"><em>Respect the cat’s choice to leave.</em></p>
</div>
```

## Explanation

Heading tags communicate hierarchy, not merely size. Paragraphs keep prose separate from headings. The ordered list conveys three steps, and strong marks the instruction label as important. Emphasis conveys stress without fake span styling. Both hacks use only the lesson’s allowed semantic tags and single-underscore `ocs_` class names. No inline styles, style blocks, or invented classes remain. The shared stylesheet supplies compatibility aliases for the spelling used in this lesson.
