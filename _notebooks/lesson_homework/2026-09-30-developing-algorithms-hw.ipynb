Prediction: 14 total hours and a 3.5-hour average. Sequencing initializes the data, iteration accumulates it, and the final division uses the list length.

```python
# CODE_RUNNER: Popcorn Hack 1

# Set up the supplied volunteer hours.
hours = [3, 5, 2, 4]
# Initialize before the loop.
total = 0
# Translate FOR EACH into Python traversal.
for h in hours:
    total += h
# Compute and display the result after traversal.
average = total / len(hours)
print(total)
print(average)
assert (total, average) == (14, 3.5)
```

## Homework Hack

Prediction: Blankets needs 12, Shampoo needs 10, total 22, with Blankets most urgent. Socks and Notebooks meet their targets exactly and are excluded.

```python
# CODE_RUNNER: Homework Hack

def restock_report(items, in_stock, target):
    # Collect each shortage and its corresponding name.
    short_items = []
    short_amounts = []
    for i in range(len(items)):
        if in_stock[i] < target[i]:
            short_items.append(items[i])
            short_amounts.append(target[i] - in_stock[i])
    # Sum shortages with an accumulator rather than sum().
    total_needed = 0
    for shortage in short_amounts:
        total_needed += shortage
    # Combine the result with an empty check before finding a maximum.
    if len(short_items) == 0:
        print("Fully stocked!")
        return [], 0, None
    # Track the position of the largest shortage; first item wins ties.
    largest_index = 0
    for i in range(1, len(short_amounts)):
        if short_amounts[i] > short_amounts[largest_index]:
            largest_index = i
    # Print each shortage and the combined total and urgency report.
    print("Restock needed:")
    for i in range(len(short_items)):
        print(f"{short_items[i]}: need {short_amounts[i]}")
    print("Total items needed:", total_needed)
    print(f"Most urgent: {short_items[largest_index]} (short by {short_amounts[largest_index]})")
    return short_items, total_needed, short_items[largest_index]

# Initialize the official supply-closet dataset.
items = ["Toothbrushes", "Blankets", "Socks", "Shampoo", "Notebooks"]
in_stock = [40, 8, 25, 5, 30]
target = [30, 20, 25, 15, 30]
# Run and check the combined report.
assert restock_report(items, in_stock, target) == (["Blankets", "Shampoo"], 22, "Blankets")
```

## Tests

An exactly stocked closet and an empty closet both report Fully stocked. Equal shortages select the first item; counts below, at, and above a seat limit distinguish strict from inclusive comparisons.

```python
# CODE_RUNNER: Tests

def restock_report(items, in_stock, target):
    # Collect each shortage and its corresponding name.
    short_items = []
    short_amounts = []
    for i in range(len(items)):
        if in_stock[i] < target[i]:
            short_items.append(items[i])
            short_amounts.append(target[i] - in_stock[i])
    # Sum shortages with an accumulator rather than sum().
    total_needed = 0
    for shortage in short_amounts:
        total_needed += shortage
    # Combine the result with an empty check before finding a maximum.
    if len(short_items) == 0:
        print("Fully stocked!")
        return [], 0, None
    # Track the position of the largest shortage; first item wins ties.
    largest_index = 0
    for i in range(1, len(short_amounts)):
        if short_amounts[i] > short_amounts[largest_index]:
            largest_index = i
    # Print each shortage and the combined total and urgency report.
    print("Restock needed:")
    for i in range(len(short_items)):
        print(f"{short_items[i]}: need {short_amounts[i]}")
    print("Total items needed:", total_needed)
    print(f"Most urgent: {short_items[largest_index]} (short by {short_amounts[largest_index]})")
    return short_items, total_needed, short_items[largest_index]

# Verify no-shortage and tie cases without indexing an empty list.
assert restock_report(["Socks"], [25], [25]) == ([], 0, None)
assert restock_report([], [], []) == ([], 0, None)
assert restock_report(["A", "B"], [0, 0], [2, 2]) == (["A", "B"], 4, "A")
# Test the boundary where apparently similar seat checks differ.
for signed_up in [11, 12, 13]:
    print("Seats:", signed_up, signed_up < 12, signed_up <= 12)
# Equal results alone do not imply equal side effects.
amounts = [25, 150, 40, 100, 10, 250]
original = amounts.copy()
largest = max(amounts)
assert amounts == original
amounts.sort()
print("Largest:", largest, "Sorted largest:", amounts[-1], "Order changed:", amounts != original)
```
