## Popcorn Hack 2

Prediction: 1.1 is rejected as a duplicate; changing only its spec to 1.2 is accepted. Every required field must pass (`and`), either racing category is allowed (`or`), and duplicates must be absent (`not`).

```python
# CODE_RUNNER: Popcorn Hack 2

part = {"product_name": "Replacement Flywheels and Clutch Assemblies", "category": "Auto Racing", "spec_number": "1.1", "effective_date": "Nov. 9, 2001"}
valid_categories = ["Auto Racing", "Drag Racing"]

def valid_record(record, existing_spec_numbers):
    has_name = bool(record.get("product_name", "").strip())
    valid_category = record.get("category") in ["Auto Racing", "Drag Racing"]
    has_spec = bool(record.get("spec_number", "").strip())
    has_date = bool(record.get("effective_date", "").strip())
    duplicate = record.get("spec_number") in existing_spec_numbers
    return has_name and valid_category and has_spec and has_date and not duplicate

existing_spec_numbers = ["1.1", "2.1"]
print("Duplicate case:", valid_record(part, existing_spec_numbers))
assert not valid_record(part, existing_spec_numbers)
part["spec_number"] = "1.2"
print("New spec:", valid_record(part, existing_spec_numbers))
assert valid_record(part, existing_spec_numbers)
```

### Recorded output

<pre>Duplicate case: False
New spec: True</pre>

## Popcorn Hack 3

Prediction: query 1.2 matches the clutch record; flywheel matches the first and third names. An otherwise matching Street Car record is excluded. Parentheses group either search match before applying the category filter.

```python
# CODE_RUNNER: Popcorn Hack 3

records = [
    {"product_name": "Replacement Flywheels and Clutch Assemblies", "category": "Auto Racing", "spec_number": "1.1"},
    {"product_name": "Multiple Disc Clutch Assemblies", "category": "Drag Racing", "spec_number": "1.2"},
    {"product_name": "Racing Flywheel Record", "category": "Auto Racing", "spec_number": "2.1"},
    {"product_name": "Street Flywheel", "category": "Street Car", "spec_number": "1.2"}
]
for query, expected in [("1.2", ["1.2"]), ("FLYWHEEL", ["1.1", "2.1"]), ("1", [])]:
    matches = []
    for record in records:
        name_match = query.lower() in record["product_name"].lower()
        spec_match = query == record["spec_number"]
        category_ok = record["category"] == "Auto Racing" or record["category"] == "Drag Racing"
        if (name_match or spec_match) and category_ok:
            matches.append(record["spec_number"])
    print(query, "->", matches)
    assert matches == expected
```

## Homework Hack and tests

Prediction for the supplied records: accepted, rejected category, rejected duplicate. Additional cases each remove one required field. Whitespace-only values are treated as empty.

```python
# CODE_RUNNER: Homework Hack and tests

def valid_record(record, existing_spec_numbers):
    has_name = bool(record.get("product_name", "").strip())
    valid_category = record.get("category") in ["Auto Racing", "Drag Racing"]
    has_spec = bool(record.get("spec_number", "").strip())
    has_date = bool(record.get("effective_date", "").strip())
    duplicate = record.get("spec_number") in existing_spec_numbers
    return has_name and valid_category and has_spec and has_date and not duplicate

existing_spec_numbers = ["1.1", "2.1"]
test_records = [
    {"product_name": "Multiple Disc Clutch Assemblies", "category": "Auto Racing", "spec_number": "1.2", "effective_date": "Feb. 9, 2006"},
    {"product_name": "Replacement Flywheels", "category": "Street Car", "spec_number": "3.1", "effective_date": "Jan. 1, 2026"},
    {"product_name": "Existing Flywheel Record", "category": "Drag Racing", "spec_number": "1.1", "effective_date": "Jan. 1, 2026"}
]
for record, expected in zip(test_records, [True, False, False]):
    accepted = valid_record(record, existing_spec_numbers)
    print(record["spec_number"], "accepted" if accepted else "rejected")
    assert accepted == expected
for field in ["product_name", "category", "spec_number", "effective_date"]:
    record = test_records[0].copy()
    record[field] = "   "
    print("Blank", field, "->", valid_record(record, existing_spec_numbers))
    assert not valid_record(record, existing_spec_numbers)
print("Missing record ->", valid_record({}, existing_spec_numbers))
assert not valid_record({}, existing_spec_numbers)
```
