## Lists

```python
# CODE_RUNNER: Lists Popcorn 1 - predict, run, fix the crash

foods = ["pizza", "sushi", "tacos", "ramen"]

print(foods[0])      # Prediction: pizza
print(foods[-1])     # Prediction: ramen
print(len(foods))    # Prediction: 4
print(foods[3])      # Fixed from invalid index
```

### Popcorn 1 Explanation

Python indexing starts at 0, so a list with four items has valid indexes from 0 through 3.

The last valid index is 3, so `foods[3]` prints `"ramen"`.

```python
# CODE_RUNNER: Lists Popcorn 2

numbers = [7, 6, 8, 5, 7, 9]

total = 0

for number in numbers:
    total += number

average = total / len(numbers)

print(total)
print(average)
```

### Popcorn 2 Explanation

The accumulator starts at 0 and adds each number in the list.

The total is 42. Dividing 42 by 6 gives an average of 7.0.

```python
# CODE_RUNNER: Lists Popcorn 3 - count then filter

seasons = ["winter", "summer", "fall", "summer", "spring", "summer", "fall"]

frequency = {}

for season in seasons:
    if season in frequency:
        frequency[season] += 1
    else:
        frequency[season] = 1

popular = []

for season in frequency:
    if frequency[season] >= 2:
        popular.append(season)

print(frequency)
print(popular)
```

### Popcorn 3 Explanation

The frequency dictionary counts how many times each season appears.

The filter creates a new list containing only seasons with at least 2 votes. The popular seasons are `"summer"` and `"fall"`.

## Homework Hack

```python
# CODE_RUNNER: Lists Homework - indexing, accumulator, count and filter on one dataset

votes = ["chips", "cookies", "chips", "fruit", "cookies", "chips",
         "pretzels", "cookies", "chips", "fruit", "chips", "cookies"]

prices = [2, 3, 2, 4, 3, 2, 1, 3, 2, 4, 2, 3]

# LESSON 1 - indexing
print(len(votes))
print(votes[-1])

# LESSON 2 - sum and average
total = 0

for price in prices:
    total += price

average = total / len(prices)

print("Total:", int(total))
print("Average:", int(average))

# LESSON 3 - count and filter
frequency = {}

for vote in votes:
    if vote in frequency:
        frequency[vote] += 1
    else:
        frequency[vote] = 1

print(frequency)

popular_snacks = []

for snack in frequency:
    if frequency[snack] >= 3:
        popular_snacks.append(snack)

print(popular_snacks)
```

### Homework Explanation

There are 12 votes, and the last vote is `"cookies"`.

The prices are totaled using an accumulator loop. The total is 31 and the integer average is 2.

The frequency dictionary counts how many votes each snack received.

The final list includes only snacks with at least 3 votes, so the result is `["chips", "cookies"]`.
