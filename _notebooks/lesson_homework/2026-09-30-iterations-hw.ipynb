## Popcorn Hack

For the supplied alerts with thresholds 3 and 9: predict six checked, two skipped, four action alerts, score total 22, and a critical alert found. The final 5 is never visited.

```python
# CODE_RUNNER: Popcorn Hack

def scan_alerts(alerts, safe_cutoff, critical_cutoff):
    checked_count = 0
    skipped_count = 0
    action_count = 0
    total_score = 0
    critical_found = False
    for alert in alerts:
        checked_count += 1
        if alert < safe_cutoff:
            skipped_count += 1
            print(alert, "safe: skipping")
            continue
        action_count += 1
        total_score += alert
        print(alert, "action needed")
        if alert >= critical_cutoff:
            critical_found = True
            print("Critical alert: stop scanning")
            break
    print("Checked:", checked_count)
    print("Skipped:", skipped_count)
    print("Actions:", action_count)
    print("Total score:", total_score)
    print("Critical found:", critical_found)
    return checked_count, skipped_count, action_count, total_score, critical_found

alerts = [1, 4, 2, 6, 3, 9, 5]
assert scan_alerts(alerts, 3, 9) == (6, 2, 4, 22, True)
```

## Homework Hack

My monitoring system checks school computer-lab server load percentages. Loads below 60 are safe; loads at least 90 are critical. Prediction: six checked, two skipped, four actions, total 291, critical True. The eight-value dataset includes two values after the critical load to verify early stopping.

```python
# CODE_RUNNER: Homework Hack

def scan_alerts(alerts, safe_cutoff, critical_cutoff):
    checked_count = 0
    skipped_count = 0
    action_count = 0
    total_score = 0
    critical_found = False
    for alert in alerts:
        checked_count += 1
        if alert < safe_cutoff:
            skipped_count += 1
            print(alert, "safe: skipping")
            continue
        action_count += 1
        total_score += alert
        print(alert, "action needed")
        if alert >= critical_cutoff:
            critical_found = True
            print("Critical alert: stop scanning")
            break
    print("Checked:", checked_count)
    print("Skipped:", skipped_count)
    print("Actions:", action_count)
    print("Total score:", total_score)
    print("Critical found:", critical_found)
    return checked_count, skipped_count, action_count, total_score, critical_found

server_loads = [22, 60, 38, 72, 65, 94, 81, 40]
safe_cutoff = 60
critical_cutoff = 90
print("Computer lab server load report")
assert scan_alerts(server_loads, safe_cutoff, critical_cutoff) == (6, 2, 4, 291, True)
```

## Tests and while-loop practice

Empty input, all-safe input, both threshold boundaries, and critical-first input check skip/stop ordering. The cooling loop updates temperature every pass so it terminates.

```python
# CODE_RUNNER: Tests and while-loop practice

def scan_alerts(alerts, safe_cutoff, critical_cutoff):
    checked_count = 0
    skipped_count = 0
    action_count = 0
    total_score = 0
    critical_found = False
    for alert in alerts:
        checked_count += 1
        if alert < safe_cutoff:
            skipped_count += 1
            print(alert, "safe: skipping")
            continue
        action_count += 1
        total_score += alert
        print(alert, "action needed")
        if alert >= critical_cutoff:
            critical_found = True
            print("Critical alert: stop scanning")
            break
    print("Checked:", checked_count)
    print("Skipped:", skipped_count)
    print("Actions:", action_count)
    print("Total score:", total_score)
    print("Critical found:", critical_found)
    return checked_count, skipped_count, action_count, total_score, critical_found

for readings, expected in [([], (0, 0, 0, 0, False)), ([10, 59], (2, 2, 0, 0, False)),
                           ([60, 89, 90, 100], (3, 0, 3, 239, True)), ([90, 60], (1, 0, 1, 90, True))]:
    assert scan_alerts(readings, 60, 90) == expected
temperature = 85
while temperature > 30:
    temperature -= 15
    print("Cooling:", temperature)
assert temperature == 25
```
