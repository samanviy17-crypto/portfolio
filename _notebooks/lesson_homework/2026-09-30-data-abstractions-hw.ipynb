## Homework

Each element stores one level’s collected stars in level order. Prediction: [12, 23, 10, 7], four levels, 52 stars.

```python
# CODE_RUNNER: Homework

level_stars = [12, 18, 10]
level_stars[1] += 5
level_stars.append(7)
total_stars = 0
for stars in level_stars:
    total_stars += stars
print(level_stars)
print(len(level_stars))
print(total_stars)
assert level_stars == [12, 23, 10, 7] and total_stars == 52
```

## Tests

The total-only loop works for four levels and no levels. Empty input skips the second-level update. The same traversal also totals the lesson’s original and extended examples.

```python
# CODE_RUNNER: Tests

for level_stars, expected in [([12, 23, 10, 7], 52), ([], 0), ([10, 20, 15], 45), ([10, 20, 15, 12], 57)]:
    total_stars = 0
    for stars in level_stars:
        total_stars += stars
    print("Levels:", len(level_stars), "Stars:", total_stars)
    assert total_stars == expected
# Assignment aliases a mutable list; copy() separates these numeric elements.
level_stars = [12, 18, 10]
other = level_stars
snapshot = level_stars.copy()
other[0] = 14
print("Shared:", level_stars, "Independent copy:", snapshot)
```

## Design Thinking

A maker needs to grow a game without losing scores. My goal is one collection with readable labels such as Forest, Cavern, Summit, and Harbor. I considered separate variables but used a list and traversal, so adding the hundredth level changes the data without adding another variable or sum term. The prototype handles four levels and an empty project; the empty test confirms that no invalid index is read. Updating a list still requires recalculating a previously stored total. No peer feedback is claimed.
