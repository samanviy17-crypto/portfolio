```python
# Popcorn Challenge

stars = 8
checkpoint_stars = stars
stars = stars + 4

print(stars)
print(checkpoint_stars)
```

## Popcorn Explanation

**Prediction:** The live stars will be 12 and the checkpoint stars will be 8.

**Actual Output:** 12, 8

**Explanation:** `checkpoint_stars` stores the value of `stars` at the time it is assigned. Changing `stars` later does not automatically change the checkpoint.

```python
# Homework Hack - UESL Star Counter

game_name = "Galaxy Star Quest"
reduced_motion = True

level_one_stars = 12
level_two_stars = 18
level_three_stars = 10

total_stars = level_one_stars + level_two_stars + level_three_stars

print(game_name)
print(reduced_motion)
print(total_stars)

level_two_stars = level_two_stars + 5

print(total_stars)

total_stars = level_one_stars + level_two_stars + level_three_stars

print(total_stars)
```

## Homework Explanation

**Prediction:** The original total will be 40. After adding 5 stars to level two, the stored total will still be 40. After recalculating, the total will become 45.

The total does not automatically update because `total_stars` stores the result of the earlier calculation. It must be reassigned to calculate the new total.

```python
# Test 2 - Bonus of 0

level_one_stars = 12
level_two_stars = 18
level_three_stars = 10

total_stars = level_one_stars + level_two_stars + level_three_stars
print(total_stars)

level_two_stars = level_two_stars + 0
print(total_stars)

total_stars = level_one_stars + level_two_stars + level_three_stars
print(total_stars)
```

## Design Thinking

**Player Need:** Some players may prefer fewer animations or less movement while playing.

**Goal:** Give players control over their experience while keeping track of their game progress.

**Setting Considered:** I used a Boolean variable called `reduced_motion` because the setting only needs two choices: True or False.

**Prototype:** I created a program that stores the game name, reduced-motion setting, level scores, and total stars.

**Test/Revision:** I tested the program with a bonus of 5 and a bonus of 0. The tests showed that a stored total does not automatically update when another variable changes.

**Why reduced motion should be a player choice:** Different players have different accessibility needs and preferences, so they should be able to choose whether reduced motion is turned on.
