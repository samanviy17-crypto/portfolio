## Popcorn Hack 2

The helper returns a Boolean; the outer procedure chooses a message. Prediction: Medical Emergency is priority, while Road Debris is routine in this classroom model.

```python
# CODE_RUNNER: Popcorn Hack 2

def is_priority(incident_type):
    return incident_type == "Medical Emergency"

def process_report(location, incident_type):
    priority = is_priority(incident_type)
    if priority:
        print(f"Priority report at {location}: {incident_type}")
    else:
        print(f"Routine report at {location}: {incident_type}")

process_report("Community Library", "Medical Emergency")
process_report("Oak Park", "Road Debris")
assert is_priority("Medical Emergency") is True
assert is_priority("Road Debris") is False
```

### Recorded output

<pre>Priority report at Community Library: Medical Emergency
Routine report at Oak Park: Road Debris</pre>

## Homework Hack and tests

The classroom priority rule recognizes Medical Emergency and Fire; other incident types use the routine branch. Decomposition isolates the rule so changing it does not require rewriting every response message.

```python
# CODE_RUNNER: Homework Hack and tests

def check_priority(incident_type):
    return incident_type in ["Medical Emergency", "Fire"]

def dispatch_response(location, incident_type):
    priority = check_priority(incident_type)
    if priority:
        print(f"Priority response requested at {location}: {incident_type}")
    else:
        print(f"Routine response requested at {location}: {incident_type}")

# Different arguments reuse the same two procedures.
dispatch_response("Community Library", "Medical Emergency")
dispatch_response("Oak Park", "Fire")
dispatch_response("Civic Center", "Supply Delivery")
for incident, expected in [("Medical Emergency", True), ("Fire", True), ("Supply Delivery", False), ("", False)]:
    actual = check_priority(incident)
    assert isinstance(actual, bool) and actual == expected
    print(repr(incident), "priority:", actual)
```
