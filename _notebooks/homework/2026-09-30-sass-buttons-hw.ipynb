## Popcorn Hack 2 — Modifiers and Groups

The check-in uses pill and fill; the secondary navigation action stays outlined. The local check-in button only changes this demonstration, without sending personal information.

```python
%%html
<!-- UI_RUNNER: Buttons Popcorn 2 | editor: true, id: buttons-popcorn-2 -->
<div id="preview-buttons-popcorn-2">
<div class="ocs__links">
<button type="button" class="ocs__btn accent pill fill">Complete Daily Check-In</button>
<a class="ocs__btn outline" href="https://www.mygoodbrain.org/resources" target="_blank" rel="noopener noreferrer">Save Draft Journal</a>
</div>
</div>
<script>
(function () {
const root = document.getElementById("preview-buttons-popcorn-2");
root.querySelectorAll('button').forEach(button => {
 button.addEventListener('click', () => {
  button.textContent = 'Recorded in this demo';
  button.disabled = true;
 });
});
})();
</script>
```

## Popcorn Hack 3 — Tones and Wide Layout

## Explanation and test cases

Buttons perform local actions; anchors navigate to the lesson’s resource destination. Fill emphasizes the primary action, outline distinguishes a supporting action, and alert-red marks urgency. Text labels also communicate intent, so the meaning does not depend on color alone. The copied starter’s Markdown-shaped hrefs are replaced with actual HTTPS URLs. All groups wrap at narrow widths, and controls remain keyboard accessible. The resource links lead to the resource directory, not a direct hotline call. No custom CSS, hex colors, or inline styles appear in these hacks.
