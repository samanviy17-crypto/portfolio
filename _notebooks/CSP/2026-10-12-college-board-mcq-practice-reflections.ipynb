# AP CSP College Board MCQ Practice Reflections

I reviewed my College Board multiple-choice results and identified three topics I want to improve.

## Topic 1: Algorithms and Procedures

Questions: Q21 (Comparing Robot Algorithms) and Q57 (Generate Happy from Harp and Puppy)

- What I struggled with: Both questions were incorrect. I had trouble tracing how algorithms and procedures work.
- What I learned: Algorithms follow specific steps, and procedures can organize instructions into reusable parts.
- How I can improve: I can trace each line of code and write down how variables change.
- My takeaway: Going through the algorithm step by step will help me avoid missing important details.

## Topic 2: Cloud Computing and the Internet

Question: Q64 (Cloud Computing and the Internet)

- What I struggled with: I answered this question incorrectly and need to understand cloud computing better.
- What I learned: Cloud computing allows people to access applications, storage, and computing resources through the Internet.
- How I can improve: I can review examples of cloud services and how they differ from storing files locally.
- My takeaway: Cloud computing uses remote resources instead of relying only on my own device.

## Topic 3: Loops and Iteration

Questions: Q25 (Shortening Strings by Replacing TH and IS) and Q56 (Skip Counting Integers in a Loop)

- What I struggled with: I answered both questions correctly, but they took longer than most of my other questions.
- What I learned: Loops repeat instructions, and the order of operations affects the result.
- How I can improve: I can create a small tracing table to track values after each iteration.
- My takeaway: Practicing loops will help me understand code faster and become more confident.
