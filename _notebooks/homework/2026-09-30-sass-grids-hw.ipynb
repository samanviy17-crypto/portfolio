## Popcorn Hack 2 — Size the Beakers with fr

Four values create four columns. `1fr 1fr 1fr 2fr` divides available track space into five shares: each trial receives one fifth and the control receives two fifths. Gaps are deducted before this division. This is the lesson’s explicit inline-style exception; the homework below contains no inline styles.

```python
%%html
<!-- UI_RUNNER: Grids Popcorn 2 | editor: true, id: grids-popcorn-2 -->
<div class="ocs__grid ocs__grid--standard" style="grid-template-columns: 1fr 1fr 1fr 2fr;">
<div class="ocs__grid-cell">Be<wbr>ak<wbr>er 1</div>
<div class="ocs__grid-cell">Be<wbr>ak<wbr>er 2</div>
<div class="ocs__grid-cell">Be<wbr>ak<wbr>er 3</div>
<div class="ocs__grid-cell">Con<wbr>trol Be<wbr>ak<wbr>er</div>
</div>
```

## MCQ answers and reasoning

1. **A — cols-4:** four equal tracks in a standard grid.
2. **B:** `2fr 1fr` creates two tracks; the first gets twice the second’s share.
3. **C — header:** spans the full row; wide spans only two columns.
4. **D:** an invented row class plus inline flex styling violates the shared grammar rule.

**MCQ Grids: 4/4 | answers: A,B,C,D**

## Homework Hack — Trial and Calculator

The header labels the trial, and the accent identifies Beaker 3 as the outlier. The calculator variant supplies four keypad columns. Every check is computed from the preview DOM. Choose 7, 8, and 9, then Avg to get 8.00.

```python
%%html
<!-- UI_RUNNER: Grids Homework | editor: true, id: grids-homework -->
<div id="preview-grids-homework">
<div class="ocs__grid ocs__grid--standard">
  <div class="ocs__grid-cell ocs__grid-cell--header">Temperature Trial (°C)</div>
  <div class="ocs__grid-cell">Beaker 1: 22.4</div>
  <div class="ocs__grid-cell">Beaker 2: 22.9</div>
  <div class="ocs__grid-cell ocs__grid-cell--accent">Beaker 3: 31.6</div>
</div>
<div class="ocs__grid ocs__grid--calculator" aria-label="Beaker reading keypad">
  <button type="button" class="ocs__grid-cell">7</button>
  <button type="button" class="ocs__grid-cell">8</button>
  <button type="button" class="ocs__grid-cell">9</button>
  <button type="button" class="ocs__grid-cell ocs__grid-cell--accent">Avg</button>
</div>
<p aria-live="polite" data-readings>Readings: none. Select values, then Avg.</p>
</div>
<script>
(function () {
const root = document.getElementById("preview-grids-homework");
// Check actual rendered markup, rather than displaying a hardcoded PASS list.
const grids = Array.from(root.querySelectorAll('.ocs__grid'));
const allowed = /^(ocs__card|ocs__grid|ocs__grid--(standard|calculator)|ocs__grid-cell|ocs__grid-cell--(header|accent|muted|wide)|cols-[234])$/;
const classes = Array.from(root.querySelectorAll('[class]')).flatMap(el => Array.from(el.classList));
const checks = [
 ['No inline style attributes', !root.querySelector('[style]')],
 ['No made up classes', classes.every(c => allowed.test(c))],
 ['Both roots use ocs__grid with exactly one variant', grids.length === 2 && grids.every(g => Array.from(g.classList).filter(c => c.startsWith('ocs__grid--')).length === 1)],
 ['Uses the standard variant for Grid 1', grids[0]?.classList.contains('ocs__grid--standard')],
 ['Uses the calculator variant for Grid 2', grids[1]?.classList.contains('ocs__grid--calculator')],
 ['Every box uses ocs__grid-cell', grids.every(g => Array.from(g.children).every(c => c.classList.contains('ocs__grid-cell'))) && grids.reduce((n,g) => n + g.children.length, 0) === 8],
 ['Title is a header cell', grids[0]?.firstElementChild.classList.contains('ocs__grid-cell--header')],
 ['Outlier Beaker 3 and Avg key are accent cells', grids[0]?.lastElementChild.classList.contains('ocs__grid-cell--accent') && grids[1]?.lastElementChild.classList.contains('ocs__grid-cell--accent')]
];
const list = document.createElement('ul');
list.setAttribute('data-checklist', 'grids');
checks.forEach(([label, ok]) => { const li = document.createElement('li'); li.textContent = (ok ? 'PASS: ' : 'FIX: ') + label; list.appendChild(li); });
root.prepend(list);
// Each run starts a new reading list; Avg never divides by zero.
const readings = [];
const status = root.querySelector('[data-readings]');
root.querySelectorAll('.ocs__grid--calculator button').forEach(button => {
 button.addEventListener('click', () => {
  if (button.textContent === 'Avg') {
   status.textContent = readings.length ? 'Average: ' + (readings.reduce((a,b) => a+b, 0) / readings.length).toFixed(2) : 'Add a reading first.';
  } else { readings.push(Number(button.textContent)); status.textContent = 'Readings: ' + readings.join(', '); }
 });
});
})();
</script>
```
