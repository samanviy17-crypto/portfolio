Prediction: total 135, difference 90, product 250, quotient 10.0. `/` produces a floating-point quotient.

## Popcorn Hack 2

Prediction: 12 spots booked, 88 left, and 44.0 per room. Multiplication happens before subtraction.

```python
# CODE_RUNNER: Math Expressions - practice 4

total_monthly_capacity = 100
spots_requested_per_group = 4
sessions_booked = 3

total_spots_taken = spots_requested_per_group * sessions_booked
remaining_capacity = total_monthly_capacity - total_spots_taken
users_per_room = remaining_capacity / 2

print("Total Spots Taken:", total_spots_taken)
print("Remaining Capacity:", remaining_capacity)
print("Users per Room:", users_per_room)
```

Remaining Capacity: 88
Users per Room: 44.0</pre>

## Explanation and tests

The first-month fee is added once. The daily average uses the monthly session cost, without the enrollment fee. Expected homework results: 105, 420, 495, 14.0. The lesson URL currently returns 404 directly; the indexed official lesson was used to verify its requirements.

```python
# CODE_RUNNER: Explanation and tests

def cost_report(price, sessions, weeks, fee):
    weekly = price * sessions
    monthly = weekly * weeks
    return weekly, monthly, monthly + fee, monthly / 30
for inputs, expected in [((35, 3, 4, 75), (105, 420, 495, 14.0)),
                         ((35, 0, 4, 75), (0, 0, 75, 0.0)),
                         ((20, 2, 4, 0), (40, 160, 160, 160 / 30))]:
    actual = cost_report(*inputs)
    assert actual == expected
    print(inputs, "->", actual)
```
