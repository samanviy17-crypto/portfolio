The single outer container establishes a readable boundary. One card groups the invitation. A standard three-column grid replaces the handmade row, and the start time receives the accent modifier.

## Homework Hack — Parade Hub Page

The hub nests container → card → grid/cells, followed by a wrapped table and a callout. A full-row header explains the grid, an accent calls attention to parade time, and the muted cleanup item remains visible without competing with the main event. These are sample planning details.

```python
%%html
<!-- UI_RUNNER: Containers Homework | editor: true, id: containers-homework -->
<div class="ocs__container">
 <div class="ocs__card">
  <h2 class="ocs__section-title">Poway Scripps Rotary Parade</h2>
  <p class="ocs__description">A sample parade hub that brings the schedule and volunteer tasks into one readable place.</p>
  <div class="ocs__grid ocs__grid--standard cols-3">
   <div class="ocs__grid-cell ocs__grid-cell--header">Parade day overview</div>
   <div class="ocs__grid-cell ocs__grid-cell--accent">Parade: 9:00 AM on Poway Road</div>
   <div class="ocs__grid-cell">Volunteers: greet visitors and help the setup crew</div>
   <div class="ocs__grid-cell ocs__grid-cell--muted">Afterward: help the cleanup team</div>
  </div>
  <div class="ocs__table-wrap">
   <table class="ocs__table">
    <caption>Sample volunteer assignments</caption>
    <thead><tr><th scope="col">Team</th><th scope="col">Task</th></tr></thead>
    <tbody><tr><td>Welcome crew</td><td>Guide visitors to the parade route</td></tr>
    <tr><td>Cleanup crew</td><td>Collect litter after the parade</td></tr></tbody>
   </table>
  </div>
  <p class="ocs__callout">Confirm actual event details with the Rotary Club before attending.</p>
 </div>
</div>
```

## Explanation and validation

The container limits width and centers the page. The card groups related information; the grid controls columns; the table represents row-and-column data. All preview content sits in exactly one container per independent hack. The standard grid changes from three columns on desktop to two on mobile. Shared theme variables control text, borders, and accent colors. No custom class or inline style is needed. The optional custom SCSS extension is intentionally omitted so the required work uses only official grammar.
